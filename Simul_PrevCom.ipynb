```python
pv = 250000
n = 144

rates_annual = [0.0, 0.04, 0.05, 0.06]
results = []

for r_a in rates_annual:
    if r_a == 0:
        pmt = pv / n
        r_m = 0
    else:
        r_m = (1 + r_a)**(1/12) - 1
        pmt = pv * (r_m) / (1 - (1 + r_m)**(-n))
    ir = pmt * 0.10
    net = pmt - ir
    results.append((r_a, r_m, pmt, ir, net))

for r in results:
    print(f"Annual: {r[0]*100:.1f}%, Monthly: {r[1]*100:.4f}%, Gross: {r[2]:.2f}, IR(10%): {r[3]:.2f}, Net: {r[4]:.2f}")
```
